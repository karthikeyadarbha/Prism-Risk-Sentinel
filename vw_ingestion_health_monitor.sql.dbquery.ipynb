```sql
/**
  To satisfy the Regulatory Reporting and Governance pillars, you need a view that monitors the health of this ingestion. Run this SQL in Databricks to create a dynamic monitor:
**/
CREATE OR REPLACE VIEW `prism-sentinel-stream`.prism_bronze.vw_ingestion_health_monitor AS
SELECT 
    date_trunc('hour', recovery_timestamp) as observation_hour,
    count(*) as total_records,
    sum(case when is_malformed = true then 1 else 0 end) as quarantined_count,
    (sum(case when is_malformed = true then 1 else 0 end) / count(*)) * 100 as failure_rate_pct
FROM `prism-sentinel-stream`.prism_bronze.transactions_raw
GROUP BY 1
ORDER BY 1 DESC;
```
